```python
import os
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import torch
torch.cuda.empty_cache()

from ultralytics import YOLO
```

```python
def training_model():
    model = YOLO("yolov8n.pt")

    history = model.train(
        data="data/data.yaml",
        epochs=10,
        batch=4,
        device='cuda',
        workers=4,
        imgsz=1024,
        optimizer='adamw',
        lr0=0.01,
        momentum=0.9
    )
```

```python
if __name__ == "__main__":
    training_model()
```
